## Metadata discovery
Locate metadata files only at a shallow level first. Do not assume a filename or label schema until the actual files are inspected.

```python
if DATASET_ROOT.exists():
    candidates = [p for p in DATASET_ROOT.iterdir()
                  if p.is_file() and any(k in p.name.lower() for k in ["meta", "label", "annotation", "split"])]
    print("\n".join(str(p) for p in candidates) if candidates else "No obvious metadata files found at dataset root.")
else:
    print("Skipped: dataset root is unavailable.")
```
