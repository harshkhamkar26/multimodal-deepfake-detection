## CNN concepts
A convolutional layer applies learned kernels across local regions. Nonlinear activations add expressive power, while pooling or strided convolutions reduce spatial resolution. For real experiments, visualize cached dataset frames instead of synthetic examples.

```python
try:
    import torch
    import torch.nn as nn
    conv = nn.Conv2d(in_channels=3, out_channels=8, kernel_size=3, padding=1)
    sample = torch.randn(1, 3, 96, 96)
    feature_maps = conv(sample)
    print("Input:", tuple(sample.shape), "Feature maps:", tuple(feature_maps.shape))
except ImportError:
    print("Install PyTorch to run the convolution example.")
```
