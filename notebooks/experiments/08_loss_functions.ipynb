## Binary classification
For two classes represented as `0/1`, use one logit per sample with `BCEWithLogitsLoss`. It combines sigmoid and binary cross-entropy in a numerically stable operation.

```python
import torch
import torch.nn as nn

binary_logits = torch.tensor([0.8, -1.2, 1.4])
binary_targets = torch.tensor([1.0, 0.0, 1.0])
binary_loss = nn.BCEWithLogitsLoss()(binary_logits, binary_targets)
print("Binary loss:", float(binary_loss))
```

## Four-class classification
Use four raw logits per sample and integer class IDs `0..3` with `CrossEntropyLoss`. Class weights, if used for imbalance, must be estimated from training data only.

```python
multiclass_logits = torch.tensor([[1.2, 0.2, -0.1, 0.0],
                                   [0.1, 0.3, 1.1, -0.5]])
class_targets = torch.tensor([0, 2], dtype=torch.long)
multiclass_loss = nn.CrossEntropyLoss()(multiclass_logits, class_targets)
print("Multi-class loss:", float(multiclass_loss))
```

### Common mistakes
- Applying softmax before `CrossEntropyLoss`.
- Passing floating-point one-hot labels to a class-index setup.
- Deriving class weights from test labels.
- Reporting loss alone without precision, recall, F1, and confusion matrix.
