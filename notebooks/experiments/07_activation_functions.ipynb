# 07 — Activation Functions

Compare common activation functions and understand their effect on forward activations and gradients.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-6, 6, 400)
functions = {
    "ReLU": np.maximum(0, x),
    "Sigmoid": 1 / (1 + np.exp(-x)),
    "Tanh": np.tanh(x),
    "Leaky ReLU": np.where(x > 0, x, 0.01 * x),
}
for name, y in functions.items():
    plt.plot(x, y, label=name)
plt.axhline(0, linewidth=0.5)
plt.axvline(0, linewidth=0.5)
plt.legend()
plt.title("Activation functions")
plt.grid(True)
plt.show()
```

## Where they are used
ReLU/Leaky ReLU are common in convolutional feature extractors. Sigmoid is appropriate for a single-logit binary probability after training, while softmax converts multi-class logits to a probability distribution. Use logits directly with `BCEWithLogitsLoss` or `CrossEntropyLoss`; do not apply sigmoid/softmax before those losses.

```python
try:
    import torch
    import torch.nn as nn
    logits = torch.tensor([[-1.0, 0.5, 2.0]])
    print("Softmax probabilities:", nn.Softmax(dim=1)(logits))
    print("Sigmoid probabilities:", torch.sigmoid(logits))
except ImportError:
    print("Install PyTorch to run tensor examples.")
```
