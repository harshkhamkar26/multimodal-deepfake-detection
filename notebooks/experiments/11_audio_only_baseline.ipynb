Build a compact CNN baseline for log-mel spectrograms and evaluate audio-only evidence.

## Input and model
Use cached log-mel spectrograms with a consistent shape. Treat the spectrogram as a single-channel image for a compact 2D CNN baseline. Keep audio preprocessing fixed across all splits.

```python
import torch
import torch.nn as nn

class AudioOnlyClassifier(nn.Module):
    def __init__(self, num_classes=1):
        super().__init__()
        self.network = nn.Sequential(
            nn.Conv2d(1, 16, 3, padding=1), nn.ReLU(), nn.MaxPool2d(2),
            nn.Conv2d(16, 32, 3, padding=1), nn.ReLU(), nn.MaxPool2d(2),
            nn.Conv2d(32, 64, 3, padding=1), nn.ReLU(),
            nn.AdaptiveAvgPool2d(1), nn.Flatten(), nn.Linear(64, num_classes)
        )
    def forward(self, mel):
        # mel: [batch, 1, mel_bins, time_steps]
        return self.network(mel)

model = AudioOnlyClassifier()
print("Parameters:", sum(p.numel() for p in model.parameters()))
```
