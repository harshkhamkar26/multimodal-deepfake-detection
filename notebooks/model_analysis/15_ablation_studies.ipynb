## Fair comparison rules
Keep data splits, sample IDs, preprocessing, evaluation metrics, and threshold-selection policy consistent. Change one main factor at a time. Repeat important runs with multiple seeds when resources allow. Do not repeatedly tune based on test metrics.

```python
def append_ablation_result(path, experiment, seed, config, dev_f1, test_f1=None, notes=""):
    row = pd.DataFrame([{
        "experiment": experiment, "seed": seed, "config": str(config),
        "best_dev_macro_f1": dev_f1, "test_macro_f1": test_f1, "notes": notes
    }])
    header = not path.exists()
    row.to_csv(path, mode="a", header=header, index=False)
    print("Saved:", path.resolve())
# Call only with measured results from completed runs.
```
