```python
def report_classification(y_true, y_pred, labels=None, target_names=None):
    from sklearn.metrics import classification_report, confusion_matrix, accuracy_score, f1_score
    print("Accuracy:", accuracy_score(y_true, y_pred))
    print("Macro F1:", f1_score(y_true, y_pred, average="macro"))
    print(classification_report(y_true, y_pred, labels=labels, target_names=target_names, zero_division=0))
    return confusion_matrix(y_true, y_pred, labels=labels)

# Call after loading real test labels and predictions; do not use placeholder predictions.
```
